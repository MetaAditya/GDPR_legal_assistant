GATHER THE ASSISTANT DATA

```python
import json
from pathlib import Path

folder = Path("./gdpr_responses")
assistant_data=[]

for file_path in folder.glob("*.json"):
    file_dict={}
    with open(file_path, "r", encoding="utf-8") as f:
        data = json.load(f)

    

    file_dict['question']=data.get("input_query","")
    file_dict['answer']=data.get("content","")
    file_dict['execution_steps']=data.get("execution_steps",[])
    file_dict['tool_query']=data.get("tool_query","")
    file_dict['tool_content']=data.get("tool_content","")
    file_dict['question_id']=data.get("question_id","")
    

    

    assistant_data.append(file_dict)


###
question_ids=[]
for ad in assistant_data:
    question_ids.append(ad.get("question_id"))
    
######

new_data_list=[]

for quid in question_ids:
    pair=[]
    for ad in assistant_data:
        if (quid==ad.get("question_id")):
            pair.append(ad)

    new_data_list.append(pair)
  

unified_list=[]
for nd in new_data_list:
    uni_dict={}
    for ele in nd:
        if ele.get("question"):
            uni_dict["question"]=ele.get("question")
        if ele.get("answer"):
            uni_dict["answer"]=ele.get("answer")
        if ele.get("tool_content"):
            uni_dict["tool_content"]=ele.get("tool_content")
        if ele.get("execution_steps"):
            uni_dict["execution_steps"]=ele.get("execution_steps")
            
    unified_list.append(uni_dict)
  
```

```python
correctness_check_data={}
for ele in unified_list:
    if ele.get("question") and ele.get("answer"):
        correctness_check_data[ele.get("question")]=ele.get("answer")
    
correctness_check_data
```

```python
from langsmith import Client

client = Client()
dataset_name="GDPR RAG Evaluation dataset"

dataset = client.create_dataset(
    dataset_name=dataset_name
)

examples = base_data

client.create_examples(
    inputs=[{"question": x["question"]} for x in examples],
    outputs=[{"answer": x["answer"]} for x in examples],
    dataset_id=dataset.id,
)

print("Dataset created:", dataset.id)

#######################
```

CORRECTNESS EVALUATOR

```python
correctness_system_prompt="""

Rewrite the following evaluation prompt:

“You are evaluating the correctness of a GDPR legal assistant’s answer against a reference answer. Assess how well the assistant’s answer semantically and substantively matches the reference answer. Exact wording is not required; focus on whether the key legal facts, conclusions, and meaning are correct.

Return a **score between 0 and 1**, where:

* **1** = fully correct and substantively matches the reference answer
* **0** = completely incorrect or does not match the reference answer
* Values between 0 and 1 represent partial correctness or semantic similarity.

Return **only the numerical score**.”


"""
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langsmith.schemas import Example, Run


llm = ChatOpenAI(
    model="gpt-5",
    temperature=0
)


prompt = ChatPromptTemplate.from_messages([
    (
        "system",correctness_system_prompt

    ),
    (
        "human",
        """
        Question:
        {question}
        
        Reference answer:
        {reference}
        
        Assistant answer:
        {answer}
        """
    )
])


def correctness_evaluator(
    run: Run,
    example: Example
):

    question = example.inputs["question"]

    reference = example.outputs["answer"]

    answer = run.outputs.get("answer", "")


    response = llm.invoke(
        prompt.format_messages(
            question=question,
            reference=reference,
            answer=answer
        )
    )

    score = response.content.strip()
    score=float(score)
    if score>=0.7:
        score=1
    else:
        score=0

    return {
        "key": "correctness",
        "score": score
    }
```

```python
from typing import *
def gdpr_correctness_check(question: str):
    
    answers=correctness_check_data
    return answers.get(
        question,
        "I don't know."
    )

```

```python
dataset_name
```

```python
from langsmith import Client
from langsmith.evaluation import evaluate

client = Client()


def target(inputs: dict):

    question = inputs["question"]

    answer = gdpr_correctness_check(question)

    return {
        "answer": answer
    }


experiment_results = evaluate(
    target,
    data=dataset_name,
    evaluators=[
        correctness_evaluator
    ],
    experiment_prefix="GDPR-Correctness-Test"
)


print(experiment_results)
```
